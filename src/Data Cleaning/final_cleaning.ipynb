```python
import pandas as pd

data = pd.read_csv("final_daily_data.csv")
```

```python
data = data.drop(columns=['pm2_5' , 'pm2_5_lag_1d',
                    'pm2_5_lag_2d',
                    'pm2_5_lag_3d',
                    'pm2_5_lag_7d',
                    'pm2_5_roll_avg_3d',
                    'pm2_5_roll_avg_7d',
                    'temperature_2m',
                    'relative_humidity_2m',
                    'surface_pressure',
                    'precipitation',
                    'wind_speed_10m'])
```

```python
hourly_data = pd.read_csv("clean_hourly_meteo.csv")
```

```python
hourly_data["date"] = pd.to_datetime(hourly_data["date"])

df_daily = hourly_data.groupby(hourly_data["date"].dt.floor("D")).agg(
    {
        "pm10": ["mean", "min", "max", "std", "var"],
        "carbon_monoxide": ["mean", "min", "max", "std", "var"],
        "sulphur_dioxide": ["mean", "min", "max", "std", "var"],
        "ozone": ["mean", "min", "max", "std", "var"],
        "nitrogen_dioxide": ["mean", "min", "max", "std", "var"],
        "pm2_5_corrected": [ "min", "max", "std", "var"]
    }
)

# Optional: Flatten the multi-level column names for a cleaner layout
df_daily.columns = [f"{col}_{func}" for col, func in df_daily.columns]
df_daily.head()
```

```python
# 1. Reset index
df_daily = df_daily.reset_index()

# 2. Convert to datetime and strip timezone info so both match
df_daily["date"] = pd.to_datetime(df_daily["date"]).dt.tz_localize(None)
data["date"] = pd.to_datetime(data["day"]).dt.tz_localize(None)

# 3. Merge on 'date'
merged_df = pd.merge(data, df_daily, on="date", how="left")

merged_df.to_csv("Merged_cleaned_daily_data.csv")
```
